```python
from youtube_transcript_api import YouTubeTranscriptApi
from langchain_text_splitters import RecursiveCharacterTextSplitter

```

```python
from langchain_ollama import ChatOllama, OllamaEmbeddings
```

```python
from langchain_community.vectorstores import FAISS
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv
```

```python
## document loader
yt_api = YouTubeTranscriptApi()
id = 'RSB58m7Xwhg'
transcript = yt_api.fetch(id)
```

```python
transcript
```

```python
[doc.text for doc in transcript]
```

```python
transcript = " ".join([doc.text for doc in transcript])
transcript
```

```python
##splitting the text
splitter = RecursiveCharacterTextSplitter(chunk_size = 500,chunk_overlap=50)
```

```python
chunks = splitter.split_text(transcript)
```

```python
len(chunks)
chunks
```

```python
e =OllamaEmbeddings(model="nomic-embed-text")
```

```python
## vector store

vectore_store = FAISS.from_texts(chunks,e)
```

```python
vectore_store.similarity_search(query= "power?")
```

```python
##it retrieves the important documents according to the query
retriever = vectore_store.as_retriever(search_kwargs={'k':4})
retriever.invoke("what is power?")
```

```python
##augumentation
```

```python
prompt = PromptTemplate(
    template="""
You are a helpful assistant.

Answer the following question using only the provided context.

Context:
{context}

Question:
{question}
Do not say phrases like:
- "Based on the provided context"
- "According to the context"
- "It appears that"

If the context does not contain enough information to answer the question, simply say:
"I can't answer based on the provided context."
""",
    input_variables=["context", "question"]
)
```

```python
question = "what is power ?"
```

```python
context = retriever.invoke(question)
```

```python
context = " ".join(i.page_content for i in context)
```

```python
final_prompt = prompt.invoke({'context':context,'question':question})
```

```python
llm = ChatOllama(model="llama3")
```

```python
answer = llm.invoke(final_prompt)
```
